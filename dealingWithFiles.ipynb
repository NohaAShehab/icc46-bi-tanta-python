# read data from file ?


```python
try:
    # open file
    filobj =  open("users.txt", 'r')
    print(filobj)
    data = filobj.read() # read file content into one string
    print(type(data), data)
    filobj.close()
except Exception as e:
    print(e)
```

# read file content line by line ?


```python
try:
    # open file
    filobj =  open("users.txt", 'r')
    print(filobj)
    data = filobj.readlines() # read file content line by line into a list
    print(type(data), data)
    filobj.close()
except Exception as e:
    print(e)
```

```python
try:
    # open file
    filobj =  open("users.txt", 'r')
    print(filobj)
    lines = filobj.readlines() # read file content line by line into a list
    data = filobj.read()
    print(type(data), data)
    filobj.close()
except Exception as e:
    print(e)
```

seek

```python
try:
    # open file
    filobj =  open("users.txt", 'r')
    print(filobj)
    lines = filobj.readlines() # read file content line by line into a list
    filobj.seek(0)  # reset the pointer to the beginning of the file
    data = filobj.read()
    print(type(data), data)
    filobj.close()
except Exception as e:
    print(e)
```

# write content to the file

```python
"""
if you are to open file with mode w --
-> if file doesn't exist, python will try to create it
--> if exists --> will open it for writing starting the beginning of the file (removing the existing content)
"""
try:
    # open file
    fileobj =  open("new.txt", 'w')
    print(fileobj)
    fileobj.close()
except Exception as e:
    print(e)
```

```python

try:
    # open file
    fileobj =  open("new.txt", 'w')
    print(fileobj)
    fileobj.write("test")
    fileobj.write('Hello')
    fileobj.close()
except Exception as e:
    print(e)
```

```python

try:
    # open file
    fileobj =  open("new.txt", 'w')
    print(fileobj)
    fileobj.writelines(["ALi", "Ahmed", "Mohamed"])
    fileobj.close()
except Exception as e:
    print(e)
```

```python
try:
    # open file
    fileobj =  open("new.txt", 'w')
    print(fileobj)
    fileobj.writelines(["ALi\n", "Ahmed\n", "Mohamed\n"])
    fileobj.close()
except Exception as e:
    print(e)
```

# Append

```python
"""
if you are to open file with mode w --
-> if file doesn't exist, python will try to create it
--> if exists --> will open it for writing starting the end of the file (keep the existing content)
"""

try:
    # open file
    fileobj =  open("new.txt", 'a')
    print(fileobj)
    fileobj.writelines(["***\n","ALi\n", "Ahmed\n", "Mohamed\n"])
    fileobj.close()
except Exception as e:
    print(e)
```
