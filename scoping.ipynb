## Python Variable Scope
- Introduces global vs local scope.
- Sets up a `course` variable for the examples that follow.


```python
# scoping --> scopes --> defines the accessibility of variables

# any variable defined in python file ===> global variable
# any variable define in the function  ==> local variable

course = 'python'
```

### Using a global variable
- Prints the value of `course` defined in the global scope.


```python
print(course)
```

### Reassigning a global variable
- Updates `course` and prints the new value.


```python
course = 'Introduction to Python'
print(course)
```

### Local variables inside a function
- Prompts for a name and shows that `name` is local to the function.


```python
def ask_for_name():
    name = input("Enter your name: ") # define local variable inside the function
    # can be accessed only inside the function
    print(name)

ask_for_name()
print(name)
```

### Reading a global from a function
- Prints the global `year` without redefining it.


```python
year = 2026

def say_hello():
    """ from the function you can read/print the value of global variable"""
    print(f"Hello we are in year {year}")

say_hello()
```

### Shadowing a global name
- Creates a local `year` that hides the global value.


```python
# I need to modify the year
year = 2026

def say_hello():
    year = "updated" # new local variable
    print(f"Hello we are in year {year}")

say_hello()
print(year)
```

### Modifying a global variable
- Uses the `global` keyword to update `year` from inside the function.


```python
# I need to modify the year
year = 2026

def modify_year():
    """ please don't create new local variable with name year,
    use the global one"""
    global year  # will not create new variable
    year = "updated" # new local variable
    print(f"Hello we are in year {year}")


modify_year()
print(year)
```

### Local scope inside a function
- Defines a local `username` and prints it.


```python
def outerFunction():
    username= "iti"

    print(f"---- username ----{username}")

outerFunction()
```

### Inner function reading an outer variable
- Shows how a nested function can access `username` from its parent.


```python
def outer_function():
    username= "iti"
    def print_username():
        """
            you access the local variable username from inner function
        :return:
        """
        # print value of variable username
        print(f"---- username ----{username.upper()}")
    print_username()

    print(f"---- username ----{username}")

outer_function()
```

### Shadowing in nested scope
- Inner `username` hides the outer one without changing it.


```python
def outer_function():
    username= "iti" # local variable
    def modify_username():
        username= "updated username" # new local variable for the inner function
        print(f"---- username ----{username.upper()}")
    modify_username()

    print(f"---- username ----{username}")

outer_function()
```

### Global assignment from an inner function
- Declares `username` as global inside the nested function.


```python
def outer_function():
    username= "iti" # local variable
    def modify_username():
        global username # update global variable, define new variable in the global scope
        username= "updated username"
        print(f"---- username ----{username.upper()}")
    modify_username()

    print(f"---- username ----{username}")

outer_function()
```

### Modifying enclosing variables with `nonlocal`
- Updates the parent function's `username` from the inner function.


```python
def outer_function():
    username= "iti" # local variable
    def modify_username():
        "modify local variable of the parent function"
        nonlocal username # please don't create new local variable inside the function
        username= "updated username"
        print(f"---- username ----{username.upper()}")
    modify_username()

    print(f"---- username ----{username}")

outer_function()
```

### Creating a global on demand
- Uses `global` to create `new_val` if it doesn't exist.


```python
def say_hello():
    global  new_val  # IF var doesn't exist will define it.
    new_val = "iti"
    print(f"New val {new_val}")
say_hello()
print(new_val)
```
