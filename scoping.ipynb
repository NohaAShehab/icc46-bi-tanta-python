<h1 style='color:green; text-align:center'>Accessing global variables inside the function</h1>

```python
year = 2026

def say_hello():
    """ from the function you can read/print the value of global variable"""
    print(f"Hello we are in year {year}")

say_hello()
```

```python
# I need to modify the year
year = 2026

def say_hello():
    year = "updated" # new local variable
    print(f"Hello we are in year {year}")

say_hello()
print(year)
```

```python
# I need to modify the year
year = 2026

def modify_year():
    """ please don't create new local variable with name year,
    use the global one"""
    global year  # will not create new variable
    year = "updated" # new local variable
    print(f"Hello we are in year {year}")


modify_year()
print(year)
```

```python
def outer_function():
    username= "iti"
    def print_username():
        """
            you access the local variable username from inner function
        :return:
        """
        # print value of variable username
        print(f"---- username ----{username.upper()}")
    print_username()

    print(f"---- username ----{username}")

outer_function()
```

```python
def outer_function():
    username= "iti" # local variable
    def modify_username():
        username= "updated username" # new local variable for the inner function
        print(f"---- username ----{username.upper()}")
    modify_username()

    print(f"---- username ----{username}")

outer_function()
```

```python
def outer_function():
    username= "iti" # local variable
    def modify_username():
        global username # update global variable, define new variable in the global scope
        username= "updated username"
        print(f"---- username ----{username.upper()}")
    modify_username()

    print(f"---- username ----{username}")

outer_function()
```

```python
def outer_function():
    username= "iti" # local variable
    def modify_username():
        "modify local variable of the parent function"
        nonlocal username # please don't create new local variable inside the function
        username= "updated username"
        print(f"---- username ----{username.upper()}")
    modify_username()

    print(f"---- username ----{username}")

outer_function()
```
