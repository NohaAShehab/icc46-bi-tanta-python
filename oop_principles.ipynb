<div style="color:#1f77b4; font-weight:bold">Using `super()` to call the parent constructor.</div>

```python
# call parent constructor
class Person:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"Person : Hello I am {self.name}")

class Engineer(Person):
    def __init__(self,name, salary):
        super().__init__(name)
        self.salary = salary

e = Engineer("Ahmed", 115454)
print(isinstance(e, Person))
e.speak()
```

# Encapsulation

<div style="color:#1f77b4; font-weight:bold">Encapsulation overview and naming conventions for access levels.</div>

```python
"""
encapsulation one of the oop principles --> wrapping data and functionality into one container

data, functionality --> you control where can be used
 access modifiers
    public ---> can be used anywhere
    protected --> can be used inside the class or derived class
    private --> can be used only inside the class

    apply validation --> salary --> "iti" --->

    NO ACCESS MODIFIERS IN PYTHON
    use naming of variables to simulate the access modifiers ?
    [a-z] --> public
    [_] --> protected
    [__] --> private
"""
```

<div style="color:#9467bd; font-weight:bold">Encapsulation demo: public, protected, and private fields.</div>

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name # public
        self._email = email # protected
        self.__salary = salary # private

emp = Employee("Ahmed","ahmed@gmail.com", 541215)
print(emp.name)
```

```python
"""
    private --> control access .. apply logic
"""
class Employee:
    def __init__(self, name, email, salary):
        self.name = name # public
        self._email = email # protected
        self.set_salary(salary)

    @property
    def salary(self):
        return self.__salary* .9

    def get_salary(self):
        return self.__salary* .9

    def set_salary(self, salary):
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be integer and > 0")


emp2 = Employee("Ahmed","ahmed@gmail.com", 454545)
print(emp2.salary)
print(emp2.get_salary())
```

```python
"""
    private --> control access .. apply logic
"""
class Employee:
    def __init__(self, name, email, salary):
        self.name = name # public
        self._email = email # protected
        self.set_salary(salary)

    @property
    def full_name(self):
        return f'{self.name.upper()}'
    @property
    def salary(self):
        return self.__salary* .9

    @salary.setter
    def salary(self, salary):
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be integer and > 0")


    def get_salary(self):
        return self.__salary* .9

    def set_salary(self, salary):
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be integer and > 0")


emp2 = Employee("Ahmed","ahmed@gmail.com", 454545)
print(emp2.salary)
emp2.salary=555555
```
