# 🟦 Python Variables
A guided walk through variables, strings, and types.


## 🟩 Indentation and blocks
Python uses indentation to define code blocks. The `if` body must be indented.


```python
# indentation
num = 10
if num == 10:
    print("found")
```

## 🟠 String variables
You can store text in variables. Use single or double quotes.


```python
name= 'noha'
```

### 🟦 Multiple string fields
Store username, email, and a combined bio message.


```python
username= 'noha'
email = "noha@gmail.com"
bio= ("My name is Noha"
      "I works at ITI")

print(username)
print(bio)
print(email)
```

## 🟣 Multiline text and comments
Triple quotes keep line breaks. A standalone triple-quoted string can act like a comment.


```python
# multiline string
bio2 = '''My name is Noha
I works at ITI
I lives in Mansoura'''   # preserve new lines \n

print(bio2)
```

## 🔵 Types at runtime
`type()` shows the current type of a value.


```python
name = 'ahmed'
print(type(name)) # object from string class
```

### 🟧 Check another variable type
Now we inspect `num`.


```python
print(type(num))
```

## 🔴 Dynamic typing
A name can be reassigned to a value of a different type.


## 🟦 Dynamic typing example
The next cell reassigns a name to a different type.


```python
# loosely dynamically typed lang.
num = 10
print(num, type(num))
num = "Hello"  # re-assign
print(num, type(num))
```

## 🟡 Type casting
Convert values with functions like `str()` and `int()`.


```python
# change datatype of variable in runtime
# type casting

num1 =10
print(type(num1), num1)
```

### 🟨 Convert to string
`str()` turns the number into text.


```python
# convert to string
num1 = str(num1 )
print(type(num1), num1)
```

### 🟩 Convert numeric text to a number
`int()` parses digits into an integer.


```python
age = '33' # contain numeric value
print(type(age), age)
age = int(age) # cast to int
print(type(age), age)
```

### 🔴 Invalid cast example
Trying to convert non-numeric text to `int()` raises an error.


```python
username= 'ahmed'
print(type(username), username)
username = int(username)
print(type(username), username)
```

## 🟣 Augmented assignment
Operators like `+=` update a variable in place.


```python
grade = 50
print(grade, type(grade))
grade +=10  # grade = grade+10
print(grade, type(grade))
```

## 🔴 Comparisons
Equality checks return `True` or `False`.


```python
print(True==1 )
```

### 🟣 Another comparison
This shows a `False` result when values differ.


```python
print(1== 10 )
```

## 🟠 Logical operators
`and` and `or` return one of the operands.


```python
# logical operators

print("iti" and 10)
```

### 🔵 `or` returns the first truthy value
Here, the left value is already truthy.


```python
print("test" or "iti")
```

## 🟩 Truthiness and `bool()`
`bool()` shows how values are treated in conditions.


```python
# value ?? I don't if it re-present True or False /

print(bool("test"))
```

## 🔵 Conditionals with truthy values
Non-empty strings are truthy, so the `if` block runs.


```python
print(name)
if name:  #name represent True value
    print("Hii")
else: print("bye")
```

## 🟪 If / elif / else
Use `elif` to test multiple conditions in order.


```python
day='Sunday'
if day == 'Sunday':
    print("Wish you a happy week ^^ ")
elif day == "Thursday":
    print("Enjoy your weekend")
elif day == "Friday":
    print("Enjoy your Friday 02 ")
else:
    print("It seems impossible until it is done ;) ")
```
