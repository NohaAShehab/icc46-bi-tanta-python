## <span style="color:#27AE60">Indentation and blocks</span>
Python uses indentation to define code blocks. The `if` body must be indented.


```python
# indentation
num = 10
if num == 10:
    print("found")
```

```python
name= 'noha'
```

```python
username= 'noha'
email = "noha@gmail.com"
bio= ("My name is Noha"
      "I works at ITI")

print(username)
print(bio)
print(email)
```

## <span style="color:#9B51E0">Multiline text and comments</span>
Triple quotes keep line breaks. A standalone triple-quoted string can act like a comment.


```python
# multiline string
bio2 = '''My name is Noha
I works at ITI
I lives in Mansoura'''   # preserve new lines \n

print(bio2)
```

## <span style="color:#2D9CDB">Types at runtime</span>
`type()` shows the current type of a value.


```python
name = 'ahmed'
print(type(name)) # object from string class
```

```python
print(type(num))
```

## <span style="color:#F2C94C">Type casting</span>
You can convert values, for example turning an integer into a string with `str()`.


```python
# loosely dynamically typed lang.
num = 10
print(num, type(num))
num = "Hello"  # re-assign
print(num, type(num))
```

```python
# change datatype of variable in runtime
# type casting

num1 =10
print(type(num1), num1)
```

```python
# convert to string
num1 = str(num1 )
print(type(num1), num1)
```

```python
age = '33' # contain numeric value
print(type(age), age)
age = int(age) # cast to int
print(type(age), age)
```

```python
username= 'ahmed'
print(type(username), username)
username = int(username)
print(type(username), username)
```

```python
grade = 50
print(grade, type(grade))
grade +=10  # grade = grade+10
print(grade, type(grade))
```
