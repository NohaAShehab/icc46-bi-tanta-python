<div style="color:#d62728; font-weight:bold">Method signature pitfall: instance call with missing `self` parameter.</div>

```python
class Employee:
    count = 0
    def __init__(self, name, email, salary):
        self.name = name # instance variable ??
        self.email = email
        self.salary = salary
        self.__class__.count +=1

    def display_employee(self):
        print(f"{self.name} - {self.email} - {self.salary}")

    # this functionality is related to the class ??
    @classmethod
    def create_employee_by_name(cls, name):
        print(f"cls , {cls}")
        return cls(name,"e@gmail.com", 54552 )

    def cal_net_salary(salary): # Usually first parameter of a method is named 'self'
        print(salary)
        return salary*.8

ee = Employee("test", "test@gmail.com", 54552)
print(ee.cal_net_salary())
```
