```python
# function in python can return with multiple values
# function do task then return with res ?
def calculator():
    num1 = input("please input first number: ")
    num2 = input("please input second number: ")
    if num1.isdigit() and num2.isdigit():
        num1 = int(num1)
        num2 = int(num2)
        res = num1  + num2
        print(f"{num1} + {num2} = {res}")
        return num1, num2, res

    return False

r= calculator()
print(r)
```

### Manual validation with `isdigit`
- Validates digit strings before converting to integers.


```python
# do validation on my own
def sum_num(num1:int, num2:int) -> int:
    if num1.isdigit() and num2.isdigit():
        num1 = int(num1)
        num2 = int(num2)
        res = num1 + num2
        print(f"{num1} + {num2} = {res}")
        return res

sum_num(3,4)
```

### Type checking with `type`
- Returns `False` for non-integer inputs.


```python
def sum_num(num1:int, num2:int) -> int:
    if type(num1) == int and type(num2) == int:
        res = num1 + num2
        print(f"{num1} + {num2} = {res}")
        return res
    return False

sum_num(3,4)
sum_num("iti","python")
sum_num(22,"python")
```

### Function with a default parameter
- Provides a fallback value for `num2`.


```python
def sum_num(num1:int, num2:int = 10) -> int:
    if type(num1) == int and type(num2) == int:
        res = num1 + num2
        print(f"{num1} + {num2} = {res}")
        return res
    return False

sum_num(3,4)
```

### Using a default parameter
- Calls `sum_num` with only the required argument.


```python
sum_num(22)
```

### Default argument order rule
- Shows that non-default parameters must come before default ones.


```python
def sum_num(num1:int = 5, num2:int ) -> int:
    if type(num1) == int and type(num2) == int:
        res = num1 + num2
        print(f"{num1} + {num2} = {res}")
        return res
    return False

sum_num(3,4)
```

### Keyword arguments without `**kwargs`

```python
def introduce_your_self(info):
    print(info)

introduce_your_self(firstName='noha')
introduce_your_self(fullname='ahmedMohamed', work='iti')
introduce_your_self(nickName='Ahmed', age=30)
```

- Accepts any number of named arguments as a dictionary.


```python
# function accept variable number of keyword parameter ?
def introduce_your_self(**info):
    print(info)

introduce_your_self(firstName='noha')
introduce_your_self(fullname='ahmedMohamed', work='iti')
introduce_your_self(nickName='Ahmed', age=30)
```

### Mini-project: Hangman
- Uses functions, loops, and input to build a small guessing game.


```python

def get_char_index(char, word):
    indexes = []
    for index, c in enumerate(word):
        if char == c:
            indexes.append(index)

    return indexes

words = {"apple", "banana", "cherry", "kiwi"}
name = input("Please enter your name:  ")
print(f"----- Welcome {name} to HangMan ----")
# get word that I need to play
selected_word = words.pop()
guessed_word = len(selected_word)* "-"
print(f"try to guess this word {guessed_word}")
guessed_word = list(guessed_word)
selected_word = list(selected_word)
for i in range(0,7):
    char = input("Please enter your guess: ")
    if len(char) ==1 and char in selected_word:
        char_index = get_char_index(char, selected_word)
        for i in char_index:
            guessed_word[i] = char
        if guessed_word == selected_word:
            print(f"You won the game! The word was {''.join(guessed_word)}")
            break
        print(''.join(guessed_word))
    else:
        print("-----You can only enter one char at time ----")

else:
    print("-----You have lost the game! -----")
```
