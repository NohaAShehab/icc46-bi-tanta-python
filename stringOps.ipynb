### 🔵 change string content ??
This cell explains: **change string content ??**.


```python
# change string content ??
iti ='information Technology Institute'
# replace each i with $

new_iti = ""
for char in iti:
    if char=='i':
        new_iti +='$'
    else:
        new_iti += char

print(new_iti)
```

### 🔴 Replace with lowercasing
This cell runs: `new_iti = ""`
Focus: normalize to lowercase then replace `i` with `$`.


```python
new_iti = ""
for char in iti.lower():
    if char=='i':
        new_iti +='$'
    else:
        new_iti += char

print(new_iti)
```

### 🟡 Replace i or I
This cell runs: `new_iti = ""`
Focus: handle both lowercase and uppercase.


```python
new_iti = ""
for char in iti:
    if char=='i' or char=='I':
        new_iti +='$'
    else:
        new_iti += char

print(new_iti)
```
