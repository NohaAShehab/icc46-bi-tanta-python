```python
# change string content ??
iti ='information Technology Institute'
# replace each i with $

new_iti = ""
for char in iti:
    if char=='i':
        new_iti +='$'
    else:
        new_iti += char

print(new_iti)
```

```python
new_iti = ""
for char in iti.lower():
    if char=='i':
        new_iti +='$'
    else:
        new_iti += char

print(new_iti)
```

```python
new_iti = ""
for char in iti:
    if char=='i' or char=='I':
        new_iti +='$'
    else:
        new_iti += char

print(new_iti)
```

```python
# replace

new_iti2 = iti.replace("i","$")
print(new_iti2)
```

```python
new_iti2 = iti.replace("i","$", 2)
print(new_iti2)
```
