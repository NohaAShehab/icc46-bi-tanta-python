# dictionary


```python
"""
- info = ["noha", "bi", "iti" ]
- unlabeled data
"""

""" to add description to the data """

info = {
    "name":"noha",
    "track": "bi",
    "work" :"iti"
}
print(info)
```

```python
# no duplicates keys
info = {
    "name":"noha",
    "track": "bi",
    "work" :"iti",
    "name" : "Noha Shehab"
}
print(info)
```

```python
# access data using key
print(info["name"])
```

```python
# dict is mutable datatype ...
# you can modify content ??
info["track"]= "Power BI"
print(info)
```

```python
info["city"] = "Tanta"
print(info)
```

```python
""" dict is comma separated key value pair"""
# get len of dict

print(len(info))
```

```python
# check element exists in dict
print("Tanta" in info) # in operator --> check if it is existing in keys
# loop over dict
for val in info:
    print(f"val is {val}")
```

```python
for key in info:
    print(f"key is {key},value {info[key]}")
```

```python
# get dict keys
d_keys = info.keys()
print(d_keys, type(d_keys))
```

```python
print(list(d_keys))
```

```python
d_values = info.values()
print(d_values, type(d_values))
```

```python
d_items = info.items()
print(d_items, type(d_items))
```

```python
items = list(d_items)
print(items)
```

```python
for item in info:
    print(item)
```

```python
for item in info.values():
    print(item)
```

```python
for k, v in info.items():
    print(k, v)
```

```python
# enumeration ??
l = [3,234,23,True]
for i in range(len(l)):
    print(i, l[i])
```

```python
l_enum = enumerate(l)
print(l_enum) # iterable
```

```python
for index, val in l_enum:
    print(index, val)
```
