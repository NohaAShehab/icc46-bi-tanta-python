```python
# break , continue
# with loops

for i in range(10):
    print(i)
```

```python
# if i== 8 stop the loop ??
for i in range(10):
    if i == 8:
        break  # leave the loop
    print(i)
```

```python
# skip the iteration at certain condition
for i in range(10):
    if i == 8:
        continue  # skip this iteration
    print(i)
```
